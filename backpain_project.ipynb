```python
import pandas as pd

# Load the dataset from the root directory
# Assuming the last column is the 'class' variable and it's unnamed
df = pd.read_csv('/backpain.csv')

# Rename the last column to 'class' if it's not already named
if df.columns[-1] not in ['class', 'Class']:
    df.rename(columns={df.columns[-1]: 'class'}, inplace=True)
```

### Inspecting 'class' column values
Let's check the unique values and their counts in the 'class' column directly after loading and renaming, to understand the exact labels present.

```python
print("Unique values in 'class' column (after rename, before any mapping):")
display(df['class'].value_counts(dropna=False))
```

### Correlation Analysis with 'pelvic_incidence'

Now, I will calculate the correlation of all variables with 'pelvic_incidence' and identify those with a correlation greater than or equal to 0.7.

```python
# Calculate correlations with 'pelvic_incidence'
correlations = df.corr(numeric_only=True)['pelvic_incidence'].sort_values(ascending=False)

# Filter for correlations >= 0.7 (absolute value)
high_correlations = correlations[correlations.abs() >= 0.7]

print("Variables with correlation >= 0.7 (absolute value) with 'pelvic_incidence':")
display(high_correlations)
```

### Data Preprocessing: Encode 'class' variable and split data

```python
from sklearn.model_selection import train_test_split

# Inspect unique values before mapping for debugging
print("Unique values in 'class' column before mapping:")
display(df['class'].value_counts(dropna=False))

# Clean 'class' variable: convert to string, strip, then map
df['class'] = df['class'].astype(str).str.strip().str.title().map({'Abnormal': 1, 'Normal': 0})

# Define features (X) and target (y)
X = df.drop('class', axis=1)
y = df['class']

# Drop rows where target variable y is NaN. This handles cases where mapping might still result in NaNs
# This is important for `stratify=y` which cannot handle NaN values.
initial_len = len(y)
y = y.dropna()
X = X.loc[y.index] # Align X with the dropped rows from y

if len(y) < initial_len:
    print(f"Warning: Dropped {initial_len - len(y)} rows due to NaN values in the 'class' column after mapping.")

# Split the data into training and test sets (70:30 ratio)
# Using shuffle=True and stratify=y to ensure class distribution is maintained
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.30, random_state=42, shuffle=True, stratify=y)

print(f"Training data shape: {X_train.shape}")
print(f"Test data shape: {X_test.shape}")
```

### Percentage of classes in `y_test`

```python
# Calculate the percentage of 0 and 1 classes in y_test
class_counts = y_test.value_counts(normalize=True) * 100

print("Percentage of 0 and 1 classes in y_test:")
display(class_counts)
```
